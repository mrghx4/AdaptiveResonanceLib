# Portable ARTLib Kaggle Validation

Attach the `portable-artlib` Kaggle Dataset, then run this notebook before using the bundle in a competition notebook.

```python
import os
import sys
from pathlib import Path

os.environ.setdefault("MPLCONFIGDIR", "/tmp/mplconfig")

PORTABLE_ARTLIB = Path("/kaggle/input/portable-artlib")
assert PORTABLE_ARTLIB.exists(), f"Portable ART bundle not found: {PORTABLE_ARTLIB}"
sys.path.insert(0, str(PORTABLE_ARTLIB))

print(f"Portable ART bundle: {PORTABLE_ARTLIB}")
```

```python
import artlib
from artlib.elementary.FuzzyART import FuzzyART
from artlib.fusion.FusionART import FusionART
from artlib.supervised.ARTMAP import ARTMAP

print("artlib imported from:", Path(artlib.__file__).resolve())
```

```python
import numpy as np

rng = np.random.default_rng(42)
```

```python
# FuzzyART smoke test
X = rng.random((24, 3))
fuzzy = FuzzyART(0.5, 0.01, 1.0)
X_prep = fuzzy.prepare_data(X)
fuzzy.fit(X_prep, max_iter=1)
pred_fuzzy = fuzzy.predict(X_prep)
print("FuzzyART OK", pred_fuzzy.shape, "clusters=", fuzzy.n_clusters)
```

```python
# FusionART smoke test
X0 = rng.random((18, 2))
X1 = rng.random((18, 2))
fusion = FusionART(
    modules=[FuzzyART(0.5, 0.01, 1.0), FuzzyART(0.6, 0.01, 1.0)],
    gamma_values=np.array([0.5, 0.5]),
    channel_dims=[4, 4],
)
fusion_data = fusion.prepare_data([X0, X1])
fusion.fit(fusion_data, max_iter=1)
pred_fusion = fusion.predict(fusion_data)
pred_regression = fusion.predict_regression(fusion_data)
print("FusionART OK", pred_fusion.shape, pred_regression.shape, "clusters=", fusion.n_clusters)
```

```python
# ARTMAP smoke test
Xa = rng.random((20, 2))
yb = rng.random((20, 2))
artmap = ARTMAP(FuzzyART(0.5, 0.01, 1.0), FuzzyART(0.5, 0.01, 1.0))
Xa_prep, yb_prep = artmap.prepare_data(Xa, yb)
artmap.fit(Xa_prep, yb_prep, max_iter=1)
pred_artmap = artmap.predict(Xa_prep)
pred_artmap_reg = artmap.predict_regression(Xa_prep)
print("ARTMAP OK", pred_artmap.shape, pred_artmap_reg.shape)
```

```python
print("Portable ART Kaggle validation passed.")
```
